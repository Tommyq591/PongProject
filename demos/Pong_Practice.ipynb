```python
import gymnasium as gym
import ale_py
from gymnasium.utils.play import play

gym.register_envs(ale_py)

env = gym.make(
    "ALE/Pong-v5",
    render_mode="rgb_array",
    frameskip=1,                   # Show every game frame.
    repeat_action_probability=0.0, # Make keyboard control predictable.
)

try:
    play(
        env,
        keys_to_action={"w": 2, "s": 3, " ": 1},
        zoom=3,
        fps=60,
    )
finally:
    env.close()
```
